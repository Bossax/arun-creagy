## 2. Sparsity Audit (The 'Mess' Factor)
In DDPM data, a `0` value in an impact field usually means 'no data collected' rather than 'verified zero damage'. We audit the percentage of records that are effectively 'empty'.

```python
def audit_sparsity(df, cols):
    results = []
    for col in cols:
        if col not in df.columns: continue
        
        if df[col].dtype in ['int64', 'float64']:
            # 0 or NaN is sparse
            empty_count = ((df[col].isna()) | (df[col] == 0)).sum()
        else:
            # Empty string or NaN is sparse
            empty_count = ((df[col].isna()) | (df[col].astype(str).str.strip() == '')).sum()
            
        results.append({
            'Column': col,
            'Sparse Pct': (empty_count / len(df)) * 100,
            'Valid Count': len(df) - empty_count
        })
    return pd.DataFrame(results).sort_values('Sparse Pct', ascending=False)

print("Top 10 Most Sparse Columns across all groups (climate hazards only):")
all_cols = [col for sublist in groups.values() for col in sublist]
audit_sparsity(df_plot, all_cols).head(10)
```

## 7. Spatial Visualization: Chiang Rai Case Study
We map the disaster impact at the Tambon (Subdistrict) level to identify high-risk hotspots. We use a fallback matching strategy to ensure data from subdistricts with missing codes are still visualized.

```python
import geopandas as gpd

SHP_PATH = r"C:\Users\sitth\OracleWorkspace\Arun_Creagy\ψ\incubate\DCCE\CRI\data_system\data\1_silver\dopa\tambon_boundaries_enriched.shp"

print("Aggregating subdistrict statistics...")
# Normalize Subdistrict Code into a strict 6-digit DOPA tambon code (e.g., 570101).
# This prevents garbage codes (e.g., '123', blanks, NaN) from entering the join key.
df_cr_plot = df_cr_plot.copy()
df_cr_plot['subdist_code_clean'] = (
    df_cr_plot['Subdistrict Code']
      .astype(str)
      .str.extract(r'(\d+)')[0]
      .str[-6:]
)
# Keep only plausible tambon codes (6 digits) for Chiang Rai (province code 57).
valid_mask = df_cr_plot['subdist_code_clean'].str.fullmatch(r'57\d{4}', na=False)
invalid_n = int((~valid_mask).sum())
if invalid_n > 0:
    print(f"[WARN] Dropping rows with invalid Subdistrict Code for Chiang Rai join: {invalid_n}")
df_cr_plot_valid = df_cr_plot.loc[valid_mask].copy()

agg_stats = df_cr_plot_valid.groupby(['subdist_code_clean', 'Subdistrict', 'District']).agg({
    'Affected Households': 'sum',
    'Affected People': 'sum',
    'Deaths': 'sum'
}).reset_index()

print("Calculating Avg YoY Change (2560-2567)...")
yoy_pivot = df_cr_plot_valid[(df_cr_plot_valid['ปี_int'] >= 2560) & (df_cr_plot_valid['ปี_int'] <= 2567)].pivot_table(
    index=['subdist_code_clean', 'Subdistrict', 'District'], 
    columns='ปี_int', 
    values='Affected Households', 
    aggfunc='sum'
).fillna(0)

years = sorted(yoy_pivot.columns)
if len(years) > 1:
    yoy_diffs = [yoy_pivot[years[i+1]] - yoy_pivot[years[i]] for i in range(len(years)-1)]
    avg_yoy = pd.concat(yoy_diffs, axis=1).mean(axis=1).rename('Avg YoY Change')
    agg_stats = agg_stats.merge(avg_yoy, on=['subdist_code_clean', 'Subdistrict', 'District'], how='left')
else:
    agg_stats['Avg YoY Change'] = 0
agg_stats['Avg YoY Change'] = agg_stats['Avg YoY Change'].fillna(0)

print("Loading shapefile and matching geometries...")
gdf = gpd.read_file(SHP_PATH)
gdf_cr = gdf[gdf['P_NAME_T'] == 'เชียงราย'].copy()

# Matching Logic
# NOTE: For a province-level case study, it's normal for some tambons in the shapefile
# to have *no disasters recorded* in the stats window. The integrity test we want is:
#   ✅ every tambon present in the disaster statistics has a matching geometry
#   (NOT the reverse).
agg_stats['match_id'] = agg_stats['subdist_code_clean'].astype(str).str.strip()
gdf_cr['match_id'] = gdf_cr['subdist_cd'].astype(str).str.strip()

# Strict integrity check: stats → geometry coverage (no fallback)
stats_vs_geom = agg_stats[['match_id', 'Subdistrict', 'District']].drop_duplicates().merge(
    gdf_cr[['match_id', 'T_NAME_T', 'A_NAME_T']],
    on='match_id',
    how='left',
    indicator=True
)
missing_geom = stats_vs_geom[stats_vs_geom['_merge'] == 'left_only'].copy()
missing_n = len(missing_geom)
if missing_n > 0:
    print(f"[FAIL] Disaster stats contain tambons with no geometry match: {missing_n}")
    display(missing_geom.sort_values(['District', 'Subdistrict']).head(50))
    raise ValueError(f"Disaster stats contain tambons with no geometry match: {missing_n}. Fix upstream codes/cleaning; do not use name fallback.")

# Geometry join for mapping (left join from geometry is fine; fill missing stats as 0)
merged = gdf_cr.merge(agg_stats, on='match_id', how='left')
target_cols = ['Affected Households', 'Affected People', 'Deaths', 'Avg YoY Change']
merged[target_cols] = merged[target_cols].fillna(0)
print("Geographic matching complete (stats fully covered by geometry).")

```

```python
# Plotting the Maps
districts = merged.dissolve(by='A_NAME_T')

map_tasks = [
    ('Affected Households', 'ครัวเรือนที่ได้รับผลกระทบสะสม'),
    ('Affected People', 'จำนวนประชาชนที่ได้รับผลกระทบสะสม'),
    ('Deaths', 'จำนวนผู้เสียชีวิตสะสม'),
    ('Avg YoY Change', 'ค่าเฉลี่ยการเปลี่ยนแปลง YoY (ครัวเรือน)')
]

for col, title in map_tasks:
    fig, ax = plt.subplots(1, 1, figsize=(14, 14))
    
    # Plot Tambon Choropleth
    merged.plot(column=col, ax=ax, cmap='Reds', legend=True, 
                edgecolor='0.8', linewidth=0.5,
                legend_kwds={'label': title, 'orientation': "horizontal", 'shrink': 0.5, 'pad': 0.05})
    
    # Overlay District Boundaries
    districts.boundary.plot(ax=ax, color='#333333', linewidth=1.2, alpha=0.7)
    
    # Annotate Top 10 Tambons
    top_10 = merged.nlargest(10, col)
    for idx, row in top_10.iterrows():
        centroid = row['geometry'].centroid
        ax.annotate(text=row['T_NAME_T'], xy=(centroid.x, centroid.y),
                    xytext=(4, 4), textcoords="offset points",
                    fontsize=10, fontproperties=thai_font,
                    bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.6, ec='0.5'))

    ax.set_title(f"เชียงราย: {title} (พ.ศ. 2560-2567)", fontsize=18, fontproperties=thai_font)
    ax.set_axis_off()
    plt.show()
```

- `Avg YoY Change` ไม่เหมาะกับกรณีปีเดียว จึงกำหนดเป็น 0

```python


SHP_PATH = r"C:\\Users\\sitth\\OracleWorkspace\\Arun_Creagy\\ψ\\incubate\\DCCE\\CRI\\data_system\\data\\1_silver\\dopa\\tambon_boundaries_enriched.shp"

print("[2567-only] Aggregating subdistrict statistics...")
df_2567 = df_cr_plot[df_cr_plot['ปี_int'].eq(2567)].copy()

# Normalize Subdistrict Code into a strict 6-digit DOPA tambon code (e.g., 570101).
df_2567['subdist_code_clean'] = (
    df_2567['Subdistrict Code']
      .astype(str)
      .str.extract(r'(\d+)')[0]
      .str[-6:]
)
valid_mask = df_2567['subdist_code_clean'].str.fullmatch(r'57\d{4}', na=False)
invalid_n = int((~valid_mask).sum())
if invalid_n > 0:
    print(f"[2567-only][WARN] Dropping rows with invalid Subdistrict Code for Chiang Rai join: {invalid_n}")
df_2567_valid = df_2567.loc[valid_mask].copy()

agg_stats_2567 = df_2567_valid.groupby(['subdist_code_clean', 'Subdistrict', 'District']).agg({
    'Affected Households': 'sum',
    'Affected People': 'sum',
    'Deaths': 'sum'
}).reset_index()

# Avg YoY Change (2567 vs 2566) at tambon level
df_2566 = df_cr_plot[df_cr_plot['ปี_int'].eq(2566)].copy()
df_2566['subdist_code_clean'] = (
    df_2566['Subdistrict Code']
      .astype(str)
      .str.extract(r'(\d+)')[0]
      .str[-6:]
)
valid_mask_2566 = df_2566['subdist_code_clean'].str.fullmatch(r'57\d{4}', na=False)
df_2566_valid = df_2566.loc[valid_mask_2566].copy()

hh_2566 = df_2566_valid.groupby('subdist_code_clean')['Affected Households'].sum().rename('HH_2566')
hh_2567 = df_2567_valid.groupby('subdist_code_clean')['Affected Households'].sum().rename('HH_2567')

# Join HH_2566 into the 2567 table so we can decide where YoY is actually computable
agg_stats_2567 = agg_stats_2567.merge(hh_2566, left_on='subdist_code_clean', right_index=True, how='left')
agg_stats_2567 = agg_stats_2567.merge(hh_2567, left_on='subdist_code_clean', right_index=True, how='left')

# YoY normalization: percent change vs 2566 baseline
# (HH_2567 - HH_2566) / HH_2566
# Not computable when HH_2566 is missing or 0.
agg_stats_2567['Avg YoY Change'] = (agg_stats_2567['HH_2567'] - agg_stats_2567['HH_2566']) / agg_stats_2567['HH_2566']
agg_stats_2567.loc[agg_stats_2567['HH_2566'].isna(), 'Avg YoY Change'] = float('nan')
agg_stats_2567.loc[agg_stats_2567['HH_2566'].eq(0), 'Avg YoY Change'] = float('nan')

print("[2567-only] Loading shapefile and matching geometries...")
gdf = gpd.read_file(SHP_PATH)
gdf_cr = gdf[gdf['P_NAME_T'] == 'เชียงราย'].copy()

# Matching Logic (strict by code; no fallback)
agg_stats_2567['match_id'] = agg_stats_2567['subdist_code_clean'].astype(str).str.strip()
gdf_cr['match_id'] = gdf_cr['subdist_cd'].astype(str).str.strip()

# Strict integrity check: stats → geometry coverage
stats_vs_geom_2567 = agg_stats_2567[['match_id', 'Subdistrict', 'District']].drop_duplicates().merge(
    gdf_cr[['match_id', 'T_NAME_T', 'A_NAME_T']],
    on='match_id',
    how='left',
    indicator=True
)
missing_geom_2567 = stats_vs_geom_2567[stats_vs_geom_2567['_merge'] == 'left_only'].copy()
missing_n_2567 = len(missing_geom_2567)
if missing_n_2567 > 0:
    print(f"[2567-only][FAIL] Disaster stats contain tambons with no geometry match: {missing_n_2567}")
    display(missing_geom_2567.sort_values(['District', 'Subdistrict']).head(50))
    raise ValueError(f"[2567-only] Disaster stats contain tambons with no geometry match: {missing_n_2567}. Fix upstream codes/cleaning; do not use name fallback.")

# Geometry join for mapping (left join from geometry; fill missing stats as 0)
merged_2567 = gdf_cr.merge(agg_stats_2567, on='match_id', how='left')
target_cols = ['Affected Households', 'Affected People', 'Deaths', 'Avg YoY Change']
merged_2567[target_cols] = merged_2567[target_cols].fillna(0)
print("[2567-only] Geographic matching complete (stats fully covered by geometry).")
```

```python
# Plotting the Maps (2567-only)
districts_2567 = merged_2567.dissolve(by='A_NAME_T')

map_tasks_2567 = [
    ('Affected Households', 'ครัวเรือนที่ได้รับผลกระทบสะสม (2567)'),
    ('Affected People', 'จำนวนประชาชนที่ได้รับผลกระทบสะสม (2567)'),
    ('Deaths', 'จำนวนผู้เสียชีวิตสะสม (2567)'),
    ('Avg YoY Change', 'ค่าเฉลี่ยการเปลี่ยนแปลง YoY (ครัวเรือน) — N/A สำหรับปีเดียว')
]

for col, title in map_tasks_2567:
    fig, ax = plt.subplots(1, 1, figsize=(14, 14))
    merged_2567.plot(column=col, ax=ax, cmap='Reds', legend=True,
                    edgecolor='0.8', linewidth=0.5,
                    legend_kwds={'label': title, 'orientation': 'horizontal', 'shrink': 0.5, 'pad': 0.05})
    districts_2567.boundary.plot(ax=ax, color='#333333', linewidth=1.2, alpha=0.7)

    top_10 = merged_2567.nlargest(10, col)
    for idx, row in top_10.iterrows():
        centroid = row['geometry'].centroid
        ax.annotate(text=row['T_NAME_T'], xy=(centroid.x, centroid.y),
                    xytext=(4, 4), textcoords='offset points',
                    fontsize=10, fontproperties=thai_font,
                    bbox=dict(boxstyle='round,pad=0.2', fc='white', alpha=0.6, ec='0.5'))

    ax.set_title(f"เชียงราย: {title}", fontsize=18, fontproperties=thai_font)
    ax.set_axis_off()
    plt.show()
```
